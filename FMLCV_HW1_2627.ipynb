1.1 Given that the probability $P(\mathrm{cardiac\_risk}=High\_Risk)=p_{high}$, what is the probability of having exactly $k_{high}$ subjects with high cardiac risk in a population of $N$ individuals?

The probability follows a binomial distribution, assuming independent individuals with the same probability of high cardiac risk.

It can be computed using the following formula:

$$
P(X=k_{high})=\binom{N}{k_{high}}p_{high}^{k_{high}} (1-p_{high})^{N-k_{high}}
$$, 

1.2 With the result obtained in the previous item, apply the MLE method to obtain an equation to estimate the value $p_{high}$.

Starting with the binomial distribution from 1.1:

$$
P(K=k_{high}) = \frac{N!}{k_{high}!(N-k_{high})!} p_{high}^{k_{high}} (1-p_{high})^{N-k_{high}}
$$

To estimate $p_{high}$ using MLE, we treat $N$ and $k_{high}$ as known values and $p_{high}$ as the unknown parameter.

Using the definition of the binomial coefficient:

$$
\binom{N}{k_{high}} = \frac{N!}{k_{high}!(N-k_{high})!}
$$

Since we observe $k_{high}$ patients out of $N$, we treat these values as fixed and consider $p_{high}$ as the unknown parameter.

The likelihood function is therefore:

$$
\mathcal{L}(p_{high}) = \frac{N!}{k_{high}!(N-k_{high})!} p_{high}^{k_{high}} (1-p_{high})^{N-k_{high}}
$$

Since $N$ and $k_{high}$ are fixed, the binomial coefficient is constant with respect to $p_{high}$:

$$
C = \frac{N!}{k_{high}!(N-k_{high})!}
$$

Therefore, we can simplify the likelihood function to:

$$
\mathcal{L}(p_{high}) = C p_{high}^{k_{high}}(1-p_{high})^{N-k_{high}}
$$

Alternatively, assuming independent Bernoulli observations, we can express the likelihood as a product, where each $X_i$ is either 0 or 1:

$$
\mathcal{L}_{individual}(p_{high}) = \prod_{i=1}^{N} p_{high}^{X_i}(1-p_{high})^{1-X_i}
$$

Taking the natural logarithm of the binomial likelihood:

$$
\ln \mathcal{L}(p_{high}) = \ln C + k_{high}\ln(p_{high}) + (N-k_{high})\ln(1-p_{high})
$$

The logarithm of the individual likelihood can be expressed using a summation (since $\ln(a \cdot b)=\ln(a)+\ln(b)$).

We can also ignore the constant $C$ when maximizing the log-likelihood, since it does not depend on $p_{high}$ and therefore does not affect the MLE estimate.

$$
\ln \mathcal{L}_{individual}(p_{high}) = \sum_{i=1}^{N} \left[ X_i\ln(p_{high}) + (1-X_i)\ln(1-p_{high}) \right]
$$

Since:

$$
\sum_{i=1}^{N}X_i=k_{high}
$$

Both likelihood formulations give the same terms depending on $p_{high}$.

Taking the derivative of the binomial log-likelihood with respect to $p_{high}$, and noting that the derivative of $\ln C$ is zero:

$$
\frac{\partial \ln \mathcal{L}(p_{high})}{\partial p_{high}} = \frac{k_{high}}{p_{high}} - \frac{N-k_{high}}{1-p_{high}}
$$

Setting the derivative equal to zero:

$$
\frac{k_{high}}{p_{high}} - \frac{N-k_{high}}{1-p_{high}} = 0
$$

Solving for $p_{high}$:

$$
k_{high}(1-p_{high}) = p_{high}(N-k_{high})
$$

$$
k_{high}-k_{high}p_{high} = Np_{high}-k_{high}p_{high}
$$

$$
k_{high}=Np_{high}
$$

Therefore, the Maximum Likelihood Estimate is:

$$
\boxed{\hat{p}_{high}=\frac{k_{high}}{N}}
$$

To verify that the MLE estimate corresponds to a maximum, we calculate the second partial derivative of the log-likelihood:

$$
\frac{\partial^2 \ln \mathcal{L}(p_{high})}{\partial p_{high}^2} = -\frac{k_{high}}{p_{high}^2} - \frac{N-k_{high}}{(1-p_{high})^2}
$$

Since both terms are negative for $0<p_{high}<1$ and $0<k_{high}<N$, we have:

$$
\frac{\partial^2 \ln \mathcal{L}(p_{high})}{\partial p_{high}^2} < 0
$$

Therefore, the log-likelihood is strictly concave, confirming that the MLE estimate corresponds to a maximum:

$$
\boxed{\hat{p}_{high}=\frac{k_{high}}{N}}
$$

The MLE estimate of $p_{high}$ is the proportion of individuals with high cardiac risk in the observed sample.

1.4 When observing the features in the dataset, we notice that they are on different scales, for instance `systolic_bp_mmHg` has a mean value of 129.088, while `troponin_level_ng_mL` has a mean of 0.239. Write the code to standardize both the train and test sets. 

*Hint: Be careful to not leak information from the test set!!*

```python
import numpy as np
print("Mean value for Systolic blood pressure:", np.mean(train_df['systolic_bp_mmHg']))
print("Mean value for Serum cardiac troponin concentration", np.mean(train_df['troponin_level_ng_mL']))

def standardize_z(X_train, X_test):

    X_std_train = X_train.copy()
    X_std_test = X_test.copy()

    for column in X_train.select_dtypes("number").columns:
        X_std_train[column] = (X_train[column] - X_train[column].mean()) / X_train[column].std()

    # Using X_trains mean and variance to avoid data leakage
    for column in X_test.select_dtypes("number").columns:
        X_std_test[column] = (X_test[column] - X_train[column].mean()) / X_train[column].std()

    return X_std_train, X_std_test

z_scaled_train, z_scaled_test = standardize_z(train_df, test_df)

print("Standardized dataset: ")
print(z_scaled_train.head())
print(z_scaled_test.head())
```


We use Z-score standardization:

$$
Z = \frac{X-\mu}{\sigma}
$$

where $\mu$ and $\sigma$ are the mean and standard deviation calculated from the training set.

Z-score standardization makes numerical features comparable by centering them around 0 with a standard deviation of 1. Unlike Min-Max scaling, it is not restricted to a fixed range and is less affected by extreme minimum and maximum values. However, Z-score is still sensitive to outliers.

We use `select_dtypes(include="number")` to select only numerical columns, excluding string-features that cant be standardized.

The same training statistics are applied to the test set to prevent data leakage.

1.5 Assuming that the likelihood probability densities follow Gaussian distributions, use MLE to estimate the needed parameters for each class. 

```python
def mean_and_varians(X):

    high_cardiac = X[X["cardiac_risk"] == "High_Risk"]
    low_cardiac = X[X["cardiac_risk"] == "Low_Risk"]

    features = X.select_dtypes(include="number").columns

    # Calculate mean vectors
    mean_high = high_cardiac[features].mean().to_numpy()
    mean_low = low_cardiac[features].mean().to_numpy()

    # Calculate covariance matrices using MLE
    covariance_high = high_cardiac[features].cov(ddof=0).to_numpy()
    covariance_low = low_cardiac[features].cov(ddof=0).to_numpy()

    return mean_high, mean_low, covariance_high, covariance_low

mean_high, mean_low, covariance_high, covariance_low = mean_and_varians(z_scaled_train)


print(f"Mean vector for class 'High_Risk': {mean_high} ")
print(f"Mean vector for class 'Low_Risk': {mean_low} ")

print(f"Covariance matrix for class 'High_Risk': {covariance_high}")
print(f"Covariance matrix for class 'Low_Risk': {covariance_low}")
```

1.7 Using the estimated values in the previous questions, evaluate the performance of the proposed model in the test set using the accuracy metric.

```python
def multivariat_gaussian(x, mean, cov):
    d = len(mean)
    diff = x - mean

    determinant = np.linalg.det(cov)
    inverse = np.linalg.inv(cov)

    coefficient = 1 / ((2 * np.pi)**(d / 2) * np.sqrt(determinant))
    exponent = -0.5 * (diff.T @ inverse @ diff)

    return coefficient * np.exp(exponent)

features = z_scaled_test.select_dtypes(include="number").columns

predictions = np.zeros(len(z_scaled_test))
answer = (z_scaled_test["cardiac_risk"] == "High_Risk").astype("int").to_numpy()

for i in range(len(z_scaled_test)):
    x = z_scaled_test[features].iloc[i].to_numpy()

    liklihood_high = multivariat_gaussian(x, mean_high, covariance_high)
    liklihood_low = multivariat_gaussian(x, mean_low, covariance_low)

    # maximum a posterior
    pos_high = liklihood_high * p_high
    pos_low = liklihood_low * p_low

    if pos_high > pos_low:
        predictions[i] = 1
    else:
        predictions[i] = 0

correct = predictions == answer
acc = correct.mean()

print(f"Accuracy in the test set: {acc}")
```

In linear regression, we standardize the target variable (y) to improve numerical stability and make gradient descent easier to optimize.

Linear regression minimizes the Mean Squared Error (MSE):

$
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$

For example, when predicting house prices, an error of €100,000 gives:

$
(500000-400000)^2=10^{10}
$

Large target values can produce large gradients, making it harder to choose a suitable learning rate. If the learning rate is too high, parameter updates may become unstable.

By standardizing the target:

$
y_{\text{scaled}}=\frac{y-\mu_y}{\sigma_y}
$

we reduce the scale of the errors and gradients, making optimization easier.

This is particularly useful for models trained using gradient-based optimization, where parameters are updated iteratively. In contrast, least squares method can compute the optimal linear regression directly without gradient descent. Therefore, standardizing the target is generally not necessary for least squares, since changing its scale does not improve the underlying model.

2.2 Assuming the target variable can be described as a linear combination as the features, i.e., $$\hat{y} = \theta_0 + age\times\theta_1 + bmi\times\theta_2 + bp\times\theta_3 + ldl \times \theta_4 + hdl \times \theta_5$$

obtain the solution vector $\hat{\theta}$ that correspond to the parameters that minimize the mean squared error between your estimations $\hat{Y}$ and the ground truth observations $Y$. Evaluate the mean squared error achieved by the proposed model.

```python
def normal_eq_ols(X, Y):
    theta0 = np.ones(X.shape[0])
    X = np.column_stack((theta0, X))

    A = X.T @ X

    # Invertible check
    if not np.linalg.matrix_rank(A) == A.shape[0]:
        raise ValueError("Matrix is not invertible")

    theta_hat = np.linalg.inv(X.T @ X) @ X.T @ Y # Normal equation

    Y_hat = X @ theta_hat
    MSE = np.mean((Y_hat - Y)**2)

    return theta_hat, MSE

X = std_train[["age", "bmi", "bp", "ldl_cholesterol", "hdl_cholesterol"]].values
Y = std_train[["target"]].values

theta_hat, MSE_ols = normal_eq_ols(X, Y)

print("Linear regression (non-regularized)")
print("theta_hat:", theta_hat)

print(f"Mean squared error: {MSE_ols}")
```

2.3 In an attempt to improve the model's performance, we decide to use a add a new variable to the problem, *total_cholesterol*, which is defined as:

Will the performance of the previous model be improved by adding this new variable? If needed, describe the biological reason of why including this variable may or may not be beneficial, as long as simulations and metrics that support your answer. 

Adding total_cholesterol will not improve the model’s performance because it is a linear combination of two existing features:
$$\mathrm{total\_cholesterol} = \mathrm{ldl\_cholesterol} +\mathrm{hdl\_cholesterol}$$

This creates perfect multicollinearity, meaning that the feature matrix contains linearly dependent columns.

In the previous exercise, we used the normal equation:

$$
\hat{\theta}=(X^TX)^{-1}X^TY
$$

This requires $$X^TX$$ to be invertible, which means that it must have full rank:

$$
\operatorname{rank}(X^TX)=n
$$

For example, consider:

$$
A=
\begin{bmatrix}
1 & 2\\
2 & 4
\end{bmatrix}
$$

The second column is twice the first, meaning the columns are linearly dependent. Therefore:

$$
\operatorname{rank}(A)=1<2
$$

The matrix is not invertible.

The same problem occurs when adding total_cholesterol, since it is the sum of LDL and HDL. The updated feature matrix has 7 columns, including the intercept, but at most 6 are linearly independent:

$$
\operatorname{rank}(X^TX)\leq6<7
$$

Therefore, the normal equation cannot be solved using ordinary matrix inversion. 
AS `theta_hat, MSE = normal_eq_ols(X, Y)` returned ValueError since the matrix is not invertible.

Instead, we can use ridge regression.
Ridge regression introduces a regularization term that makes the matrix invertible:

$$
\hat{\theta}=(X^TX+\lambda I)^{-1}X^TY
$$

However, adding total_cholesterol does not provide new information, since LDL and HDL are already included in the model. Biologically, LDL and HDL have different roles, so keeping them as separate features is more informative than using only their sum.

The optimal MSE for ordinary least squares remains unchanged:

$$
MSE_{\text{original}}=MSE_{\text{updated}}
$$

This can be verified by comparing both models using a least-squares solver for $OLS$ and $Ridge$. However, ridge regression may produce a different MSE because it introduces regularization. As shown in the next code part

Therefore, adding total_cholesterol does not improve the original model’s predictive performance.


```python
def normal_eq_ridge(X, Y, lambda_=0.1):
    theta0 = np.ones(X.shape[0])
    X = np.column_stack((theta0, X))

    I = np.eye(X.shape[1])
    I[0, 0] = 0

    theta_hat = np.linalg.inv(X.T @ X + lambda_ * I) @ X.T @ Y

    Y_hat = X @ theta_hat
    MSE = np.mean((Y_hat - Y)**2)

    return theta_hat, MSE

# Add total cholesterol before standardization
train_updated = train_df.copy()
test_updated = test_df.copy()

train_updated["total_cholestrol"] = (
    train_updated["ldl_cholesterol"] + train_updated["hdl_cholesterol"]
)

test_updated["total_cholestrol"] = (
    test_updated["ldl_cholesterol"] + test_updated["hdl_cholesterol"]
)

# Use the function from task 1.4
std_train_updated, std_test_updated = standardize_z(
    train_updated, test_updated
)


X = std_train_updated[["age", "bmi", "bp", "ldl_cholesterol", "hdl_cholesterol", "total_cholestrol"]].values
Y = std_train_updated[["target"]].values



# theta_hat, MSE_ridge = normal_eq_ols(X, Y) ValueError
theta_hat, MSE_ridge = normal_eq_ridge(X, Y)
print(f"MSE_ols = {MSE_ols}, MSE_ridge = {MSE_ridge}")
```
