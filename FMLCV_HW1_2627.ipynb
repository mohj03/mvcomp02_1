1.1 Given that the probability $P(\mathrm{cardiac\_risk}=High\_Risk)=p_{high}$, what is the probability of having exactly $k_{high}$ subjects with high cardiac risk in a population of $N$ individuals?

The probability follows a binomial distribution, assuming independent individuals with the same probability of high cardiac risk.

It can be computed using the following formula:

$$
P(X=k_{high})=\binom{N}{k_{high}}p_{high}^{k_{high}} (1-p_{high})^{N-k_{high}}
$$,

### 1.2 Maximum Likelihood Estimation (MLE)
Starting with the binomial distribution from 1.1:
$$
P(X=k_{high}) =
\frac{N!}{k_{high}!(N-k_{high})!}
p_{high}^{k_{high}}
(1-p_{high})^{N-k_{high}}
$$
To estimate $p_{high}$ using MLE, we treat $N$ and $k_{high}$
as known values and $p_{high}$ as the unknown parameter.
The likelihood function is therefore:
$$
\mathcal{L}(p_{high}) =
\frac{N!}{k_{high}!(N-k_{high})!}
p_{high}^{k_{high}}
(1-p_{high})^{N-k_{high}}
$$
Since $N$ and $k_{high}$ are fixed, the binomial coefficient
is constant with respect to $p_{high}$:
$$
C = \frac{N!}{k_{high}!(N-k_{high})!}
$$
Thus:
$$
\mathcal{L}(p_{high}) =
C p_{high}^{k_{high}}(1-p_{high})^{N-k_{high}}
$$
Taking the liklihood gives us, where $X_{i}$ si an independet result:
$$
\mathcal{L}_{individual}(p_{high}) =
\prod_{i=1}^{N}
p_{high}^{X_i}(1-p_{high})^{1-X_i}
$$
Taking the natural logarithm of the binomial:
$$
\ln (p_{high}) =
\ln C
+ k_{high}\ln(p_{high})
+ (N-k_{high})\ln(1-p_{high})
$$

Therfore,

The logarithm of the individual likelihood can be
expressed using a summation (since $ln(a * b) = ln(a) + ln(b)$),
We can also look away from the constant $C$ when maximizing the log-likelihood, 
since it does not depend on $p_{high}$ and therefore does not affect the MLE estimate.
$$
\ln \mathcal{L}_{individual}(p_{high}) =
\sum_{i=1}^{N}
\left[
X_i\ln(p_{high})+
(1-X_i)\ln(1-p_{high})
\right]
$$
Since:
$$
\sum_{i=1}^{N}X_i=k_{high}
$$
Both likelihood formulations give the same terms
depending on $p_{high}$.
Taking the partial derivative of the binomial
log-likelihood with respect to $p_{high}$:
$$
\frac{\partial \ln \mathcal{L}(p_{high})}
{\partial p_{high}}
=
\frac{k_{high}}{p_{high}}
-
\frac{N-k_{high}}{1-p_{high}}
$$
Because $C$ is constant, its derivative is zero.
Therefore:
$$
\frac{\partial \ln \mathcal{L}(p_{high})}
{\partial p_{high}}
=
\frac{k_{high}}{p_{high}}
-
\frac{N-k_{high}}{1-p_{high}}
$$
Setting the derivative equal to zero:
$$
\frac{k_{high}}{p_{high}}
-
\frac{N-k_{high}}{1-p_{high}}
=0
$$
Solving for $p_{high}$:
$$
k_{high}(1-p_{high})
=
p_{high}(N-k_{high})
$$
$$
k_{high}-k_{high}p_{high}
=
Np_{high}-k_{high}p_{high}
$$
$$
k_{high}=Np_{high}
$$
Therefore, the Maximum Likelihood Estimate is:
$$
\boxed{\hat{p}_{high}=\frac{k_{high}}{N}}
$$

To verify that the MLE estimate corresponds to a maximum,
we calculate the second partial derivative of the log-likelihood:

$$
\frac{\partial^2 \ln \mathcal{L}(p_{high})}{\partial p_{high}^2}
=
-\frac{k_{high}}{p_{high}^2}
-\frac{N-k_{high}}{(1-p_{high})^2}
$$

Since both terms are negative for $0<p_{high}<1$
and $0<k_{high}<N$, we have:

$$
\frac{\partial^2 \ln \mathcal{L}(p_{high})}{\partial p_{high}^2}<0
$$

Therefore, the log-likelihood is strictly concave,
confirming that the MLE estimate corresponds to a maximum:

$$
\boxed{\hat{p}_{high}=\frac{k_{high}}{N}}
$$

The MLE estimate of $p_{high}$ is the proportion of
individuals with high cardiac risk in the observed population.

### 1.4 Standardize
We use Z-score standardization:

$$
Z = \frac{X-\mu}{\sigma}
$$

where $\mu$ and $\sigma$ are the mean and standard deviation calculated from the training set.

Z-score standardization makes numerical features comparable by centering them around 0 with a standard deviation of 1. Unlike Min-Max scaling, it is not restricted to a fixed range and is less affected by extreme minimum and maximum values. However, Z-score is still sensitive to outliers.

We use `select_dtypes(include="number")` to select only numerical columns, excluding string-features that cant be standardized.

The same training statistics are applied to the test set to prevent data leakage.

```python
def mean_and_varians(X):

    high_cardiac = X[X["cardiac_risk"] == "High_Risk"]
    low_cardiac = X[X["cardiac_risk"] == "Low_Risk"]

    features = X.select_dtypes(include="number").columns

    # Calculate mean vectors
    mean_high = high_cardiac[features].mean().to_numpy()
    mean_low = low_cardiac[features].mean().to_numpy()

    # Calculate covariance matrices using MLE
    covariance_high = high_cardiac[features].cov(ddof=0).to_numpy()
    covariance_low = low_cardiac[features].cov(ddof=0).to_numpy()

    return mean_high, mean_low, covariance_high, covariance_low

mean_high, mean_low, covariance_high, covariance_low = mean_and_varians(z_scaled_train)


print(f"Mean vector for class 'High_Risk': {covariance_low} ")
print(f"Mean vector for class 'Low_Risk': {mean_low} ")

print(f"Covariance matrix for class 'High_Risk': {covariance_high}")
print(f"Covariance matrix for class 'Low_Risk': {covariance_low}")
```

1.7 Using the estimated values in the previous questions, evaluate the performance of the proposed model in the test set using the accuracy metric.

```python
def multivariat_gaussian(x, mean, cov):
    d = len(mean)
    diff = x - mean

    determinant = np.linalg.det(cov)
    inverse = np.linalg.inv(cov)

    coefficient = 1 / ((2 * np.pi)**(d / 2) * np.sqrt(determinant))
    exponent = -0.5 * (diff.T @ inverse @ diff)

    return coefficient * np.exp(exponent)

features = z_scaled_test.select_dtypes(include="number").columns

predictions = np.zeros(len(z_scaled_test))
answer = (z_scaled_test["cardiac_risk"] == "High_Risk").astype("int").to_numpy()

for i in range(len(z_scaled_test)):
    x = z_scaled_test[features].iloc[i].to_numpy()

    liklihood_high = multivariat_gaussian(x, mean_high, covariance_high)
    liklihood_low = multivariat_gaussian(x, mean_low, covariance_low)

    # maximum a posterior
    pos_high = liklihood_high * p_high
    pos_low = liklihood_low * p_low

    if pos_high > pos_low:
        predictions[i] = 1
    else:
        predictions[i] = 0

correct = predictions == answer
acc = correct.mean()

print(f"Accuracy in the test set: {acc}")
```
